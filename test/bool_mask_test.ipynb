```python
def sample(max_val, sample_size):
    return torch.randperm(max_val)[:sample_size]

rows = (
    torch.stack([sample(4, 2) for _ in range(2)])
    * 4
)
cols = (
    torch.stack([sample(4, 2) for _ in range(2)])
    * 4
)
rows, cols
```

```python
B = 2
num_patches = 2
patch_size = 4
rows = rows.repeat_interleave(patch_size**2, dim=1) + torch.arange(patch_size).repeat(num_patches * patch_size)
cols = cols.repeat_interleave(patch_size**2, dim=1) + torch.arange(patch_size).repeat_interleave(patch_size).repeat(num_patches)
rows, cols
```

```python
row_idxs = torch.arange(start=0, end=16, step=4).repeat(16 // 4)
col_idxs = torch.arange(start=0, end=16, step=4).repeat_interleave(16 // 4)
rows, cols
```
